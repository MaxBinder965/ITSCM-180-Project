```python
def profitability_analyzer():
    print("Business Profitability Analyzer")

    # Get business information
    business_name = input("Enter business name: ")

    revenue = float(input("Enter total revenue: $"))
    expenses = float(input("Enter total expenses: $"))

    # Calculate profit
    profit = revenue - expenses

    # Calculate profit margin
    if revenue > 0:
        profit_margin = (profit / revenue) * 100
    else:
        profit_margin = None  # undefined when there is no revenue

    # Display results
    print("\n=====Business Report =====")
    print("Business:", business_name)
    print("Revenue: $", revenue)
    print("Expenses: $", expenses)
    print("Net Profit: $", profit)

    if profit_margin is None:
        print("Profit Margin: N/A (no revenue)")
    else:
        print("Profit Margin:", round(profit_margin, 2), "%")

    # Evaluate business performance
    if profit > 0:
        print("Status: The business is profitable!")
    elif profit == 0:
        print("Status: The business is breaking even.")
    else:
        print("Status: The business is operating at a loss.")

# Run the analyzer
profitability_analyzer()
```
